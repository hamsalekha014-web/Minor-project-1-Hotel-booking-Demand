 8.Detect and Handle Outliers

    Use suitable visualizations to identify outliers.


```python
df_clean = df.copy()

print("Shape before outlier removal:", df_clean.shape)
plt.figure(figsize=(8,5))

plt.boxplot(df_clean["adr"].dropna())

plt.title("Box Plot of Average Daily Rate")
plt.ylabel("ADR")

plt.show()
outlier_columns = ["adr"]

keep_rows = pd.Series(True, index=df_clean.index)

for column in outlier_columns:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    column_mask = df_clean[column].between(
        lower_bound,
        upper_bound
    )

    print(f"{column}: {(~column_mask).sum()} outlier(s)")

    keep_rows &= column_mask

df_clean = df_clean.loc[keep_rows].reset_index(drop=True)

print("Shape after outlier removal:", df_clean.shape)
```

    Compare the data before and after handling outliers


```python
print("Shape before outlier removal:", df.shape)
print("Shape after outlier removal:", df_clean.shape)
rows_removed = len(df) - len(df_clean)

print("Rows removed:", rows_removed)
```
